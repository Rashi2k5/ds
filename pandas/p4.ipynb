```python
import pandas as pd
```

```python
var = pd.DataFrame({"A" : [1,2,3,4], "B"  : [4,5,6,7]})
var["c"] = var["A"] + var["B"]
print(var)
```

```python
var["c"] = var["A"] - var["B"]
print(var)
```

```python
var["c"] = var["A"] * var["B"]
print(var)
```

```python
var["c"] = var["A"] / var["B"]
print(var)
```
