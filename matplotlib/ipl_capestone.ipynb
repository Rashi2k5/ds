```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

1.which team won the first matches

```python
match_wins = df['match_winner'].value_counts()
```

```python
sns.barplot(y = match_wins .index,x = match_wins.values,palette='viridis')
plt.title("MOST MATCH WINS BY TEAM")
```

toss decision trend


```python
df.head()
```

```python
sns.countplot(x = df['toss_decision'],palette='rainbow')
plt.title("toss decision trends")
```

toss winners vs match winners

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()

percentage = (count*100)/df.shape[0]
percentage.round(2)

```

how teams win?


```python
sns.countplot(x = df['won_by'])
plt.title("won by")
```

most player of the match awarded

```python
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x = count.values, y = count.index,palette = 'mako')
plt.title("man of the match")
```

top scorer

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind = 'barh')
```

10 best bowling figures

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wickets']  = df['highest_wickets'].astype(int)
top_bowler = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False)
top_bowler.plot(kind = 'barh')
```

most matches played by venue


```python
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(y = venue_count.index,x = venue_count.values,palette="rainbow")
```
