```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.xlabel("chacha")
plt.ylabel("bhatija")
plt.title("chacha with bivi")
```

```python
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x*x,y*7)
plt.subplot(2,2,4)
plt.plot(x,y*2)
```
